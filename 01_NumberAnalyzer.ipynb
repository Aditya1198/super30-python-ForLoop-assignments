Number Analyzer using for loop

Take a number N from the user. Print all numbers from 1 to N, identify whether each number is even or odd, and finally display the total count of even and odd numbers.

```python
# Take a number from the user
N = int(input("Enter a number N: "))

even_count = 0
odd_count = 0

# Check numbers from 1 to N
for number in range(1, N + 1):

    if number % 2 == 0:
        print(number, "is Even")
        even_count += 1
    else:
        print(number, "is Odd")
        odd_count += 1

# Display the totals
print("\nTotal Even Numbers:", even_count)
print("Total Odd Numbers:", odd_count)
```
