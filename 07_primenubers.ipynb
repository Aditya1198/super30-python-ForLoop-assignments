Prime Number Finder

Ask the user for a starting and ending number. Print all prime numbers within that range using nested loops.

```python
start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

print("Prime numbers:")

for number in range(start, end + 1):

    if number < 2:
        continue

    is_prime = True

    # Inner loop checks whether the number has any divisor
    for divisor in range(2, number):
        if number % divisor == 0:
            is_prime = False
            break

    if is_prime:
        print(number)
```
