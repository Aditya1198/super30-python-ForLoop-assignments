Create Your Own sum() Function

Write

def my_sum(numbers)

It should accept a list of numbers and return their sum without using Python's built-in sum().

```python
def my_sum(numbers):
    total = 0

    for number in numbers:
        total = total + number

    return total


numbers = [10, 20, 30, 40, 50]

result = my_sum(numbers)

print("Sum:", result)
```
