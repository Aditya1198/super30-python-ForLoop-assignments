Reusable Number Analysis Function

Create a function

def analyze_number(number): ...

It should determine whether the number is positive/negative/zero, even/odd, and prime/not prime. Return the results rather than only printing them.

```python
def analyze_number(number):
    # Positive, negative, or zero
    if number > 0:
        sign = "Positive"
    elif number < 0:
        sign = "Negative"
    else:
        sign = "Zero"

    # Even or odd
    if number % 2 == 0:
        parity = "Even"
    else:
        parity = "Odd"

    # Prime or not prime
    if number <= 1:
        prime_status = "Not Prime"
    else:
        prime_status = "Prime"

        for i in range(2, int(number ** 0.5) + 1):
            if number % i == 0:
                prime_status = "Not Prime"
                break

    return sign, parity, prime_status


# Take input from user
number = int(input("Enter a number: "))

# Call the function
result = analyze_number(number)

# Display the result
print("Number is:", result[0])
print("Number is:", result[1])
print("Number is:", result[2])
```
