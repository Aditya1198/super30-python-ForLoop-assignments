Shopping Cart using Functions

Create functions for add_item(), remove_item(), view_cart(), calculate_total(), and checkout(). Keep the application running using a while loop until checkout or exit.

```python
cart = []


def add_item():
    name = input("Enter item name: ")
    price = float(input("Enter item price: "))
    
    if price < 0:
        print("Price cannot be negative.")
        return

    item = {
        "name": name,
        "price": price
    }

    cart.append(item)
    print(name, "added to cart.")


def remove_item():
    if len(cart) == 0:
        print("Your cart is empty.")
        return

    view_cart()

    item_number = int(input("Enter item number to remove: "))

    if item_number >= 1 and item_number <= len(cart):
        removed_item = cart.pop(item_number - 1)
        print(removed_item["name"], "removed from cart.")
    else:
        print("Invalid item number.")


def view_cart():
    if len(cart) == 0:
        print("Your cart is empty.")
        return

    print("\n--- Your Cart ---")

    for i, item in enumerate(cart, start=1):
        print(i, ".", item["name"], "- ₹", item["price"])

    print("Total: ₹", calculate_total())


def calculate_total():
    total = 0

    for item in cart:
        total = total + item["price"]

    return total


def checkout():
    if len(cart) == 0:
        print("Your cart is empty. Nothing to checkout.")
        return False

    print("\n--- Checkout ---")
    view_cart()

    confirm = input("Do you want to checkout? (yes/no): ")

    if confirm.lower() == "yes":
        print("Payment successful!")
        print("Thank you for shopping!")
        return True
    else:
        print("Checkout cancelled.")
        return False


# Main application
while True:

    print("\n===== Shopping Cart =====")
    print("1. Add Item")
    print("2. Remove Item")
    print("3. View Cart")
    print("4. Calculate Total")
    print("5. Checkout")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_item()

    elif choice == "2":
        remove_item()

    elif choice == "3":
        view_cart()

    elif choice == "4":
        print("Total: ₹", calculate_total())

    elif choice == "5":
        if checkout():
            break

    elif choice == "6":
        print("Exiting shopping cart. Goodbye!")
        break

    else:
        print("Invalid choice. Please try again.")
```
