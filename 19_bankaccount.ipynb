Bank Account Mini Application

Create functions for deposit(), withdraw(), check_balance(), and transaction_history(). Use a while loop to keep the banking application running. Prevent withdrawals when sufficient balance is unavailable.

```python
balance = 0
transactions = []


def deposit():
    global balance

    amount = float(input("Enter amount to deposit: "))

    if amount <= 0:
        print("Deposit amount must be greater than 0.")
        return

    balance = balance + amount
    transactions.append("Deposited: ₹" + str(amount))

    print("₹", amount, "deposited successfully.")


def withdraw():
    global balance

    amount = float(input("Enter amount to withdraw: "))

    if amount <= 0:
        print("Withdrawal amount must be greater than 0.")
        return

    if amount > balance:
        print("Insufficient balance.")
        print("Available balance: ₹", balance)
        return

    balance = balance - amount
    transactions.append("Withdrawn: ₹" + str(amount))

    print("₹", amount, "withdrawn successfully.")


def check_balance():
    print("Current balance: ₹", balance)


def transaction_history():
    if len(transactions) == 0:
        print("No transactions found.")
        return

    print("\n--- Transaction History ---")

    for transaction in transactions:
        print(transaction)


# Main banking application
while True:

    print("\n===== Bank Account =====")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Transaction History")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        deposit()

    elif choice == "2":
        withdraw()

    elif choice == "3":
        check_balance()

    elif choice == "4":
        transaction_history()

    elif choice == "5":
        print("Thank you for using our banking application.")
        break

    else:
        print("Invalid choice. Please try again.")
```
